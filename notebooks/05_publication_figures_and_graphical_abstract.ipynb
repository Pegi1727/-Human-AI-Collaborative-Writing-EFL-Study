# Notebook 05: Publication Figures and Graphical Abstract

### Study: Generative AI in EFL Academic Writing, Self-Regulated Learning (SRL), and AI Literacy
This notebook creates high-resolution, publication-quality figures representing the core empirical findings of the study:
1. **Pre-to-Post Trajectories**: Interaction line plots illustrating the divergence between Experimental and Control groups.
2. **Distributional Changes**: Boxplots and violin distributions across groups.
3. **Adjusted Effect Sizes & Forest Plot**: Summary of effect sizes ($\eta_p^2$ and Cohen's $d$).
4. **Graphical Abstract**: A multi-panel overview suitable for journal submission.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure publication style
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['axes.edgecolor'] = '#333333'
plt.rcParams['axes.linewidth'] = 0.8
plt.rcParams['figure.dpi'] = 300

# Load data
df = pd.read_csv('/mnt/data/Hypothetical_EFL_AI_Study_Data.csv')
print(f"Data loaded successfully with {len(df)} subjects.")
```

## 1. Figure 1: Pre-to-Post Trajectory Interaction Plots
Visualizing baseline equivalence and post-intervention divergences across Academic Writing, SRL, and AI Literacy.


```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
domains = [
    ('Academic Writing (0-100)', 'Writing_Pre', 'Writing_Post'),
    ('Self-Regulated Learning (1-5)', 'SRL_Pre', 'SRL_Post'),
    ('AI Literacy (1-5)', 'AI_Pre', 'AI_Post')
]

colors = {'Experimental': '#1f77b4', 'Control': '#d62728'}

for ax, (title, pre_col, post_col) in zip(axes, domains):
    for group_name in ['Experimental', 'Control']:
        grp_data = df[df['Group'] == group_name]
        mean_pre, se_pre = grp_data[pre_col].mean(), grp_data[pre_col].sem()
        mean_post, se_post = grp_data[post_col].mean(), grp_data[post_col].sem()
        
        ax.errorbar(['Pre-test', 'Post-test'], [mean_pre, mean_post], 
                    yerr=[se_pre, se_post],
                    marker='o', markersize=8, linewidth=2.2, capsize=4,
                    label=group_name, color=colors[group_name])
    
    ax.set_title(title, fontsize=13, fontweight='bold', pad=12)
    ax.set_ylabel('Mean Score (+/- SE)', fontsize=11)
    ax.legend(title='Condition', frameon=True)
    ax.grid(True, linestyle=':', alpha=0.6)

plt.suptitle('Figure 1: Pre-to-Post Intervention Trajectories Across Study Outcomes', fontsize=15, fontweight='bold', y=1.03)
plt.tight_layout()
plt.show()
```

## 2. Figure 2: Post-Test Distributions (Box + Strip Plots)
Examines individual variance, ceiling effects, and distribution spread between groups.


```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

post_vars = [
    ('Writing_Post', 'Post-Test Academic Writing'),
    ('SRL_Post', 'Post-Test Self-Regulated Learning'),
    ('AI_Post', 'Post-Test AI Literacy')
]

palette = {'Experimental': '#4c72b0', 'Control': '#c44e52'}

for ax, (var, title) in zip(axes, post_vars):
    sns.boxplot(x='Group', y=var, data=df, ax=ax, palette=palette, width=0.45, boxprops=dict(alpha=0.7), showmeans=True,
                meanprops={"marker":"s","markerfacecolor":"white", "markeredgecolor":"black"})
    sns.stripplot(x='Group', y=var, data=df, ax=ax, color='black', alpha=0.4, jitter=0.2, size=5)
    ax.set_title(title, fontsize=12, fontweight='bold')
    ax.set_xlabel('')
    ax.set_ylabel('Score')

plt.suptitle('Figure 2: Distributional Comparison of Post-Intervention Measures', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## 3. Figure 3: Effect Size Forest Plot ($\eta_p^2$ and Adjusted Cohen's $d$)


```python
# Summary effect sizes extracted from ANCOVA models
effects_data = pd.DataFrame({
    'Outcome': ['AI Literacy', 'Self-Regulated Learning', 'Academic Writing'],
    'Cohen_d': [1.88, 1.34, 1.15],
    'CI_lower': [1.45, 0.95, 0.78],
    'CI_upper': [2.31, 1.73, 1.52],
    'Partial_Eta_Sq': [0.52, 0.36, 0.28]
})

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 4.5), gridspec_kw={'width_ratios': [1.4, 1]})

# Cohen's d Forest plot
y_pos = np.arange(len(effects_data))
ax1.errorbar(effects_data['Cohen_d'], y_pos, 
             xerr=[effects_data['Cohen_d'] - effects_data['CI_lower'], effects_data['CI_upper'] - effects_data['Cohen_d']],
             fmt='o', color='#1b5e20', ecolor='#1b5e20', elinewidth=2, capsize=5, markersize=8)
ax1.axvline(0, color='grey', linestyle='--', linewidth=1)
ax1.axvline(0.8, color='orange', linestyle=':', label="Large Effect (d=0.8)")
ax1.set_yticks(y_pos)
ax1.set_yticklabels(effects_data['Outcome'], fontsize=11, fontweight='bold')
ax1.set_xlabel("Cohen's d (Adjusted Effect Size) with 95% CI", fontsize=11)
ax1.set_title("Standardized Mean Differences (ANCOVA Adjusted)", fontsize=12, fontweight='bold')
ax1.legend(loc='lower right')

# Partial Eta Squared Bar plot
ax2.barh(y_pos, effects_data['Partial_Eta_Sq'], color='#3949ab', alpha=0.85, height=0.5)
ax2.axvline(0.14, color='red', linestyle=':', label=r"Large Effect ($\eta_p^2 \geq 0.14$)")
ax2.set_yticks(y_pos)
ax2.set_yticklabels([])
ax2.set_xlabel(r"Partial Eta Squared ($\eta_p^2$)", fontsize=11)
ax2.set_title(r"Variance Explained ($\eta_p^2$)", fontsize=12, fontweight='bold')
ax2.set_xlim(0, 0.7)
ax2.legend(loc='lower right')

for i, v in enumerate(effects_data['Partial_Eta_Sq']):
    ax2.text(v + 0.02, i, f"{v:.2f}", va='center', fontweight='bold')

plt.suptitle('Figure 3: Effect Size Magnitudes across Study Domains', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## 4. Figure 4: Graphical Abstract
An integrative visual summary suitable for scientific publication highlights and graphical abstracts.


```python
fig = plt.figure(figsize=(16, 9))
gs = fig.add_gridspec(2, 3, height_ratios=[1, 1.2])

# Panel A: Study Design Schema
ax0 = fig.add_subplot(gs[0, :])
ax0.axis('off')
box_props = dict(boxstyle='round,pad=0.8', facecolor='#e8f4f8', edgecolor='#0288d1', linewidth=1.5)
ax0.text(0.1, 0.5, "Quasi-Experimental Design
N = 120 EFL Undergraduates
(Pre-Test Assessment)", 
         ha='center', va='center', bbox=box_props, fontsize=12, fontweight='bold')
ax0.annotate('', xy=(0.28, 0.65), xytext=(0.20, 0.55),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8))
ax0.annotate('', xy=(0.28, 0.35), xytext=(0.20, 0.45),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8))

box_exp = dict(boxstyle='round,pad=0.8', facecolor='#c8e6c9', edgecolor='#2e7d32', linewidth=1.5)
box_ctrl = dict(boxstyle='round,pad=0.8', facecolor='#ffcdd2', edgecolor='#c62828', linewidth=1.5)
ax0.text(0.48, 0.72, "Experimental Condition (n=60)
GenAI Process Writing & Feedback Intervention", 
         ha='center', va='center', bbox=box_exp, fontsize=11, fontweight='bold')
ax0.text(0.48, 0.28, "Control Condition (n=60)
Traditional Instructor Process Writing Instruction", 
         ha='center', va='center', bbox=box_ctrl, fontsize=11, fontweight='bold')

ax0.annotate('', xy=(0.75, 0.5), xytext=(0.68, 0.72),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8))
ax0.annotate('', xy=(0.75, 0.5), xytext=(0.68, 0.28),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8))

box_post = dict(boxstyle='round,pad=0.8', facecolor='#fff9c4', edgecolor='#f57f17', linewidth=1.5)
ax0.text(0.88, 0.5, "Post-Test ANCOVA
Significant Gains in:
• Writing (d = 1.15)
• SRL (d = 1.34)
• AI Literacy (d = 1.88)", 
         ha='center', va='center', bbox=box_post, fontsize=11, fontweight='bold')
ax0.set_title("Panel A: Methodological Architecture and Intervention Workflow", fontsize=13, fontweight='bold', loc='left')

# Panels B, C, D: Domain comparisons
domains_summary = [
    ('Writing Performance', 'Writing_Pre', 'Writing_Post', 0),
    ('Self-Regulated Learning', 'SRL_Pre', 'SRL_Post', 1),
    ('AI Literacy', 'AI_Pre', 'AI_Post', 2)
]

for title, pre, post, idx in domains_summary:
    ax = fig.add_subplot(gs[1, idx])
    for grp in ['Experimental', 'Control']:
        sub = df[df['Group'] == grp]
        ax.plot(['Pre', 'Post'], [sub[pre].mean(), sub[post].mean()], marker='o', linewidth=2.5, label=grp, color=colors[grp])
    ax.set_title(f"Panel B{idx+1}: {title}", fontsize=12, fontweight='bold')
    ax.set_ylabel('Mean Score')
    ax.legend()
    ax.grid(True, linestyle=':', alpha=0.6)

plt.suptitle('Graphical Abstract: Generative AI Assistance in EFL Academic Writing Pedagogy', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```
