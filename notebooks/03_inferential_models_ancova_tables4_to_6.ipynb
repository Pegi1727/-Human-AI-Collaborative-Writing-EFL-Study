# Notebook 03: Inferential Statistical Models (ANCOVA) - Tables 4, 5, and 6

### Study: Generative AI in EFL Academic Writing, Self-Regulated Learning (SRL), and AI Literacy
This notebook conducts the primary inferential statistical analysis using Analysis of Covariance (ANCOVA) across the three study outcomes:
1. **Writing Performance** (Post-test controlled for Pre-test) -> **Table 4**
2. **Self-Regulated Learning (SRL)** (Post-test controlled for Pre-test) -> **Table 5**
3. **AI Literacy** (Post-test controlled for Pre-test) -> **Table 6**

Each model evaluates the main effect of `Group` (Experimental vs. Control) adjusting for baseline covariate scores, calculating Type III Sum of Squares, F-statistics, p-values, and partial eta squared ($\eta_p^2$) effect sizes.


```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Load dataset
df = pd.read_csv('/mnt/data/Hypothetical_EFL_AI_Study_Data.csv')
print(f"Dataset successfully loaded: {df.shape[0]} rows, {df.shape[1]} columns")
df.head()
```

## 1. Table 4: ANCOVA for Academic Writing Performance
- **Dependent Variable (DV)**: `Writing_Post`
- **Covariate (CV)**: `Writing_Pre`
- **Between-Subjects Factor (IV)**: `Group`


```python
# Fit ANCOVA Model for Writing Performance
model_writing = ols('Writing_Post ~ Writing_Pre + C(Group)', data=df).fit()
table4 = sm.stats.anova_lm(model_writing, typ=3)

# Calculate Partial Eta Squared: SS_effect / (SS_effect + SS_residual)
table4['eta_sq_partial'] = table4['sum_sq'] / (table4['sum_sq'] + table4.loc['Residual', 'sum_sq'])
table4['eta_sq_partial'] = table4['eta_sq_partial'].apply(lambda x: f"{x:.4f}" if pd.notnull(x) else "")

print("=== TABLE 4: ANCOVA Summary for Academic Writing Performance ===")
print(table4)
```

## 2. Table 5: ANCOVA for Self-Regulated Learning (SRL)
- **Dependent Variable (DV)**: `SRL_Post`
- **Covariate (CV)**: `SRL_Pre`
- **Between-Subjects Factor (IV)**: `Group`


```python
# Fit ANCOVA Model for SRL
model_srl = ols('SRL_Post ~ SRL_Pre + C(Group)', data=df).fit()
table5 = sm.stats.anova_lm(model_srl, typ=3)

# Calculate Partial Eta Squared
table5['eta_sq_partial'] = table5['sum_sq'] / (table5['sum_sq'] + table5.loc['Residual', 'sum_sq'])
table5['eta_sq_partial'] = table5['eta_sq_partial'].apply(lambda x: f"{x:.4f}" if pd.notnull(x) else "")

print("=== TABLE 5: ANCOVA Summary for Self-Regulated Learning (SRL) ===")
print(table5)
```

## 3. Table 6: ANCOVA for AI Literacy
- **Dependent Variable (DV)**: `AI_Post`
- **Covariate (CV)**: `AI_Pre`
- **Between-Subjects Factor (IV)**: `Group`


```python
# Fit ANCOVA Model for AI Literacy
model_ai = ols('AI_Post ~ AI_Pre + C(Group)', data=df).fit()
table6 = sm.stats.anova_lm(model_ai, typ=3)

# Calculate Partial Eta Squared
table6['eta_sq_partial'] = table6['sum_sq'] / (table6['sum_sq'] + table6.loc['Residual', 'sum_sq'])
table6['eta_sq_partial'] = table6['eta_sq_partial'].apply(lambda x: f"{x:.4f}" if pd.notnull(x) else "")

print("=== TABLE 6: ANCOVA Summary for AI Literacy ===")
print(table6)
```

## 4. Synthesis of Inferential Results across Tables 4-6
Below is a consolidated summary of the treatment effects across all three dimensions.


```python
summary_ancova = pd.DataFrame([
    {
        'Outcome Measure': 'Academic Writing',
        'F (Group)': table4.loc['C(Group)', 'F'],
        'p-value': table4.loc['C(Group)', 'PR(>F)'],
        'Partial Eta^2': float(table4.loc['C(Group)', 'eta_sq_partial']),
        'Significance': 'p < 0.001' if table4.loc['C(Group)', 'PR(>F)'] < 0.001 else f"{table4.loc['C(Group)', 'PR(>F)']:.4f}"
    },
    {
        'Outcome Measure': 'Self-Regulated Learning',
        'F (Group)': table5.loc['C(Group)', 'F'],
        'p-value': table5.loc['C(Group)', 'PR(>F)'],
        'Partial Eta^2': float(table5.loc['C(Group)', 'eta_sq_partial']),
        'Significance': 'p < 0.001' if table5.loc['C(Group)', 'PR(>F)'] < 0.001 else f"{table5.loc['C(Group)', 'PR(>F)']:.4f}"
    },
    {
        'Outcome Measure': 'AI Literacy',
        'F (Group)': table6.loc['C(Group)', 'F'],
        'p-value': table6.loc['C(Group)', 'PR(>F)'],
        'Partial Eta^2': float(table6.loc['C(Group)', 'eta_sq_partial']),
        'Significance': 'p < 0.001' if table6.loc['C(Group)', 'PR(>F)'] < 0.001 else f"{table6.loc['C(Group)', 'PR(>F)']:.4f}"
    }
])
display(summary_ancova)
```
